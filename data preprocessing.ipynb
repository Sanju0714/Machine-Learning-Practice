```python
# import libraries

import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv('ipl_data.csv')
```

```python
df.head()
```

## Categorical - Data encoding 

Label Encoding - each unique category is replaced with a simple number like (0,1,2)

```python
le=LabelEncoder()
```

```python
# fit() → learns categories

# transform() → converts categories into binary columns
```

```python
df['team_encoded']=le.fit_transform(df['team'])
```

```python
df[['team','team_encoded']].drop_duplicates()
```

## One - Hot Encoding 

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder()

encoded=ohe.fit_transform(df[['team']])
print(ohe.get_feature_names_out())
```

```python
## fit - learn from data
## transform - whatever is learning in the fit will be applied 
```

```python
ohe=OneHotEncoder(sparse_output=False)

encoded=ohe.fit_transform(df[['team']])

ohe_df=pd.DataFrame(encoded, columns=ohe.get_feature_names_out())

print(ohe_df)
```

```python
ohe=OneHotEncoder(sparse_output=False)

encoded=ohe.fit_transform(df[['team']])

ohe_df=pd.DataFrame(encoded, columns=ohe.get_feature_names_out())

res=pd.concat([df,ohe_df],axis=1)

#res = pd.concat([df['team'],ohe_df],axis=1)
print(res.head(10))
```

## Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder 

order = [['Loss','Tie','Win']]

oe = OrdinalEncoder(categories=order)

df['outcome_encoded']=oe.fit_transform(df[['match_result']])

print(df[['match_result','outcome_encoded']].drop_duplicates())
```
